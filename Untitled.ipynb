# Statistical Analysis of Car Listings (Turbo.az Dataset)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

df = pd.read_csv('/content/cars.csv')
df.head()
```

```python
df.info()
df.isnull().sum()

# Correcting the column names for dropping NaNs
df = df.dropna(subset=['price_x', 'production_year', 'engine_displacement_num', 'kilometrage_num'])

print("\nDataFrame info after dropping NaNs:")
df.info()
print("\nNaN values after dropping NaNs:")
df.isnull().sum()
```

## Hypotheses

1. Is there a difference in price between manual and automatic cars? → t-test  
2. Does car price depend on city? → chi-square  
3. Does price differ across engine sizes? → ANOVA  

```python
# Display unique values in 'transmission' to verify
print(df['transmission'].unique())

# Correcting the filter conditions using the actual values from the 'transmission' column
manual = df[df['transmission'] == 'Mexaniki']['price_azn']
auto = df[df['transmission'] == 'Avtomat']['price_azn']

# Perform t-test only if both series are not empty
if not manual.empty and not auto.empty:
    t_stat, p_value = stats.ttest_ind(manual, auto, equal_var=False) # Assuming unequal variances
    print(f"T-statistic: {t_stat:.2f}")
    print(f"P-value: {p_value:.3f}")
else:
    print("One or both transmission types have no data for price, cannot perform t-test.")
```

### Conclusion
If p < 0.05 → price differs between transmission types.

```python
df['price_category'] = pd.qcut(df['price_azn'], q=3, labels=['low','mid','high'])

table = pd.crosstab(df['city_az'], df['price_category']) # Using 'city_az' after renaming

chi2, p, _, _ = stats.chi2_contingency(table)

print(f"P-value for Chi-square test: {p:.3f}")
```

### Conclusion
If p < 0.05 → price depends on city.

```python
# Group prices by engine volume for ANOVA test
# Ensure 'engine_volume_liters' is numeric
df['engine_volume_liters'] = pd.to_numeric(df['engine_volume_liters'], errors='coerce')
df.dropna(subset=['engine_volume_liters', 'price_azn'], inplace=True)

groups = [group['price_azn'] for name, group in df.groupby('engine_volume_liters')]

# Perform ANOVA only if there is more than one group and each group has enough data
if len(groups) > 1 and all(len(g) > 0 for g in groups):
    f_stat, p_value = stats.f_oneway(*groups)
    print(f"F-statistic: {f_stat:.2f}")
    print(f"P-value: {p_value:.3f}")
else:
    print("Not enough groups or data points to perform ANOVA test.")
```

### Conclusion
If p < 0.05 → price differs by engine size.

```python
sns.boxplot(x='transmission', y='price_azn', data=df)
plt.show()

sns.countplot(x='city_az', hue='price_category', data=df)
plt.show()

sns.boxplot(x='engine_type_text', y='price_azn', data=df)
plt.show()
```
