Hypothesis testing prior to parametric testing

```python
import pandas as pd
df = pd.read_csv('/content/cars.csv', on_bad_lines='skip')
display(df.head())
```

```python
print(df.columns)
```

```python
from scipy.stats import shapiro

# Remove NaN values from the 'price_x' column before performing the Shapiro-Wilk test
price_data = df['price_x'].dropna()

stat, p = shapiro(price_data)

print("Shapiro test:")
print("Statistic =", stat)
print("p-value =", p)

if p > 0.05:
    print("Данные распределены нормально")
else:
    print("Данные НЕ распределены нормально")
```

```python
from scipy.stats import levene

# Filter out NaN values from 'Marka' and 'price_x' first
df_filtered = df[['Marka', 'price_x']].dropna()

# Use 'Marka' for grouping, assuming it's a suitable categorical column
# Ensure there are at least two unique values for the test
unique_marka = df_filtered['Marka'].unique()

if len(unique_marka) < 2:
    print("Not enough unique 'Marka' values to perform Levene's test.")
else:
    # Take the first two unique brands for comparison
    group1_name = unique_marka[0]
    group2_name = unique_marka[1]

    group1 = df_filtered[df_filtered['Marka'] == group1_name]['price_x']
    group2 = df_filtered[df_filtered['Marka'] == group2_name]['price_x']

    stat, p = levene(group1, group2)

    print("Levene test (comparing '" + str(group1_name) + "' and '" + str(group2_name) + "'):")
    print("Statistic =", stat)
    print("p-value =", p)

    if p > 0.05:
        print("Дисперсии равны")
    else:
        print("Дисперсии НЕ равны")
```

## Assumption Checking

To apply parametric tests, the normality of the distribution and the equality of variances were assessed.

The Shapiro–Wilk test results indicated that the data distribution is (not) normal.

Levene’s test showed that the variances are (equal / unequal), a factor taken into account when selecting the statistical test.
